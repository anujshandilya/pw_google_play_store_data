```python
df = df.drop_duplicates()
```

```python
df_copy = df.copy()
```

```python
df_copy.reset_index(drop = True, inplace = True)
```

```python
df_copy[~df_copy.Reviews.str.isnumeric()]
```

```python
df_copy = df_copy.drop(df_copy.index[9990])
```

```python
df_copy[~df_copy.Reviews.str.isnumeric()]
```

```python
df_copy.Reviews = df_copy.Reviews.astype(int)
```

```python
df_copy.dtypes
```

```python
df_copy.Size.unique()
```

```python
def convert_size(item):
    if str(item)[-1] == 'M':
        res = float(str(item).replace("M",""))
        res*=1024
        return res
    elif str(item)[-1] == 'K':
        res = float(str(item).replace("K",''))
        return res
    else:
        return str(np.NaN)
df_copy.Size = df_copy.Size.apply(convert_size)
```

```python
df_copy.Size.unique()
```

```python
df_copy.Size = df_copy.Size.astype("float")
```

```python
df_copy.dtypes
```

```python
df_copy.Installs = df_copy.Installs.str.replace("+","").str.replace(",","")
```

```python
df_copy.Installs = df_copy.Installs.astype(int)
```

```python
df_copy.dtypes
```

```python
df_copy.Price.unique()
```

```python
df_copy.Price = df_copy.Price.str.replace("$","")
df_copy.Price = df_copy.Price.astype(float)
```

```python
df_copy.dtypes
```

```python
df_copy['Last Updated']
```

```python
df_copy['Last Updated'] = pd.to_datetime(df_copy['Last Updated'])
```

```python
df_copy.dtypes
```

```python
df_copy['Day'] = df_copy['Last Updated'].dt.day
df_copy['Month'] = df_copy['Last Updated'].dt.month
df_copy['Year'] = df_copy['Last Updated'].dt.year
```

```python
df_copy.drop("Last Updated", axis = 1, inplace = True)
```

```python
df_copy.head()
```

```python
df_copy['Android Ver'].unique()
```

```python
df_copy['Android Ver'] = df_copy['Android Ver'].str.replace(" and up","").str.replace("Varies with device","")
```

```python
df_copy['Android Ver'].unique()
```

```python
df_copy['Android Ver'] = df_copy['Android Ver'].str.extract(r'(\d+\.\d)')[0]
```

```python
df_copy['Android Ver'].unique()
```

```python
df_copy['Android Ver'] = df_copy['Android Ver'].astype("float")
```

```python
df_copy.dtypes
```

```python
df_copy[df_copy.App.duplicated()]
```

```python
df_copy = df_copy.drop_duplicates(subset=['App'],keep='first')
```

```python
df_copy[df_copy.App.duplicated()]
```

```python
#what are top 5 installed category apps in each type? answer>> game app of free

cat = df_copy.groupby(['Category', 'Type'])['Installs'].sum().sort_values(ascending = False).reset_index()
```

```python
cat
```

```python
free_cat_top_5 = cat[cat.Type == 'Free'][['Installs','Category']].head().reset_index()
free_cat_top_5
```

```python
sns.barplot(x=free_cat_top_5['Category'],y=free_cat_top_5['Installs'], data = free_cat_top_5)
plt.title('Top 5 Free Apps')
plt.show()
```

```python
paid_cat_top_5 = cat[cat.Type == 'Paid'][['Installs','Category']].head().reset_index()
paid_cat_top_5
```

```python
sns.barplot(x=paid_cat_top_5['Category'],y=paid_cat_top_5['Installs'], data = paid_cat_top_5)
plt.title('Top 5 Paid Apps')
plt.show()
```

```python
#which category app users are reviewing the most? Social Facebook
top_reviews = df_copy.groupby('Category')['Reviews'].sum().sort_values(ascending = False).reset_index().head(10)
top_reviews
```

```python
sns.barplot(x = top_reviews['Category'], y = top_reviews['Reviews'], data = top_reviews)
plt.title("Top Category based on reviews")
plt.xticks(rotation = 90)
plt.show()
```
